# Otay Mesa Industrial-Logistics Coupling - Canyu Li

**Research question.** Is industrial activity in the Otay Mesa border area more strongly
associated with ports of entry (POEs) and major freight corridors than industrial activity
in comparable non-border industrial areas of San Diego?

**Design of this notebook.**

1. **Otay Mesa is the primary case.** Kearny Mesa, Miramar and Sorrento Valley are non-border
   comparison areas, measured with the same indicators.
2. **The unit of analysis is the census tract.** Industrial *zoning* is a coarse unit: San Diego
   is spatially constrained ("city of villages" separated by canyons), industrial zoning often
   sits on land left over after other allocations, and the city contains only a handful of
   industrial zones - too few for statistics. Census tracts give hundreds of observations and
   support real statistical comparison.
3. **Activity is measured, not just designated.** Each tract carries ACS industry-of-employment
   counts (manufacturing + wholesale + transportation/warehousing), a free and reproducible
   measure of where industrial work is located.
4. **Results are read as spatial association.** The notebook reports proximity, overlap and
   relative concentration, runs correlation/OLS, and avoids causal language.

**Data and reproducibility.** The notebook reads public sources (US Census tracts and the ACS
API, SANDAG FreightViewer, and a GADM county boundary) and needs no ArcGIS license to run its
core analysis. Census tract geometry loads from the Census TIGER service, with a GitHub-hosted
mirror as an automatic fallback when the Census host is unreachable. Optional companion datasets
(Esri GeoEnrichment by tract; geocoded business points) attach through integration hooks in
Section 10.

**Credentials are entered once**, in Section 1. A Census API key is required for the ACS request;
an ArcGIS username is optional and only used if a step needs it. Both are stored in the process
environment so no later cell asks again.

## 2. Helper functions

Reusable helpers, including diagnostics that print after each spatial layer is created.
`read_esri_or_geojson_url` handles SANDAG FreightViewer files, which are Esri FeatureSet JSON
rather than GeoJSON. `download_json_to_local` fetches a remote file with `requests` and reads it
from disk, which avoids GDAL/CURL SSL issues behind some proxies. `load_tracts_with_fallback`
tries the Census TIGER host first and a GitHub-hosted Census mirror second. `load_gadm_county`
reads the uploaded GADM boundary. `style_map_axes` applies consistent cartographic styling
(title block, data-source note, simple scale bar and north arrow).

```python
def inspect_gdf(gdf, name="GeoDataFrame", n=3):
    # Quick diagnostics; run after loading any spatial layer.
    print(f"--- {name} ---")
    print("shape:", gdf.shape, "| crs:", gdf.crs)
    print("geometry types:", dict(gdf.geom_type.value_counts(dropna=False)))
    print("columns:", list(gdf.columns))
    display(gdf.head(n))

def to_target(gdf, assumed="EPSG:4326"):
    # Reproject to TARGET_CRS; set an assumed CRS first if missing.
    if gdf.crs is None:
        gdf = gdf.set_crs(assumed)
    return gdf.to_crs(TARGET_CRS)

def download_json_to_local(url, local_path, timeout=120):
    # Fetch with requests (works behind proxies that block GDAL's CURL SSL) and cache to disk.
    local_path = Path(local_path)
    if local_path.exists() and local_path.stat().st_size > 0:
        return local_path
    r = requests.get(url, timeout=timeout)
    r.raise_for_status()
    local_path.write_bytes(r.content)
    return local_path

def read_geojson_url(url, name, assumed="EPSG:4326"):
    gdf = gpd.read_file(url)
    gdf = to_target(gdf, assumed)
    print(f"[ok] {name}: {len(gdf):,} features")
    return gdf

def find_data_file(name):
    # Resolve a data file in data_processed/ first, then the working directory.
    # Some inputs are written to the working directory rather than data_processed/,
    # so both locations are checked.
    for cand in (DATA_PROCESSED / name, Path.cwd() / name, Path(name)):
        try:
            if cand.exists():
                return cand
        except Exception:
            pass
    return DATA_PROCESSED / name   # canonical path for "not found" messages

def read_esri_or_geojson_url(url, name, assumed="EPSG:4326"):
    # SANDAG FreightViewer files are Esri FeatureSet JSON, not GeoJSON.
    # Try GeoPandas/GDAL first; if that yields nothing, parse the Esri JSON by hand.
    try:
        gdf = gpd.read_file(url)
        if len(gdf):
            print(f"[ok] {name}: {len(gdf):,} features (gdal)")
            return to_target(gdf, assumed)
    except Exception:
        pass
    from shapely.geometry import LineString, MultiLineString, Polygon, MultiPolygon
    js = requests.get(url, timeout=60).json()
    feats = js.get("features", [])
    sr = js.get("spatialReference", {}) or {}
    epsg = sr.get("latestWkid") or sr.get("wkid") or 4326
    geoms, rows = [], []
    for f in feats:
        g = f.get("geometry", {}) or {}
        attrs = f.get("attributes", {}) or {}
        geom = None
        if "x" in g and "y" in g:
            geom = Point(g["x"], g["y"])
        elif "paths" in g:
            lines = [LineString(p) for p in g["paths"] if len(p) > 1]
            geom = lines[0] if len(lines) == 1 else (MultiLineString(lines) if lines else None)
        elif "rings" in g:
            rings = [Polygon(r) for r in g["rings"] if len(r) > 2]
            geom = rings[0] if len(rings) == 1 else (MultiPolygon(rings) if rings else None)
        if geom is not None:
            geoms.append(geom); rows.append(attrs)
    gdf = gpd.GeoDataFrame(rows, geometry=geoms, crs=f"EPSG:{epsg}")
    print(f"[ok] {name}: {len(gdf):,} features (esri-json)")
    return to_target(gdf, assumed)

# Census tract sources: official TIGER first, GitHub mirror second.
TRACT_PRIMARY_URL = ("https://www2.census.gov/geo/tiger/GENZ2023/shp/"
                     "cb_2023_06_tract_500k.zip")
TRACT_MIRROR_URL  = ("https://raw.githubusercontent.com/loganpowell/census-geojson/"
                     "master/GeoJSON/500k/2020/06/tract.json")

def load_tracts_with_fallback(county_fips="073"):
    # 1) official Census TIGER cartographic boundary tracts (current vintage)
    try:
        ca = gpd.read_file(TRACT_PRIMARY_URL)
        src = "Census TIGER cb_2023 (primary)"
    except Exception as e:
        print("[fallback] Census TIGER host unreachable:", type(e).__name__, "-> GitHub mirror")
        local = download_json_to_local(TRACT_MIRROR_URL, DATA_RAW / "ca_tracts_mirror.json")
        ca = gpd.read_file(local)
        src = "GitHub census-geojson mirror (2020 500k)"
    sd = ca[ca["COUNTYFP"] == county_fips].copy()
    sd["GEOID"] = sd["GEOID"].astype(str)
    sd = sd[sd.geometry.notna() & ~sd.geometry.is_empty].copy()
    sd = to_target(sd)
    sd["area_sqmi"] = sd.geometry.area / (FT_PER_MILE ** 2)
    print(f"[tracts] source: {src} | San Diego County tracts: {len(sd):,}")
    return sd[["GEOID", "NAME", "geometry", "area_sqmi"]] if "NAME" in sd.columns \
           else sd[["GEOID", "geometry", "area_sqmi"]]

def load_gadm_county(name_1="California", name_2="SanDiego"):
    # County administrative boundary from the uploaded GADM level-2 file.
    # The southern edge of San Diego County is the US-Mexico international border,
    # which is why this boundary is useful context for the Otay Mesa maps.
    if not Path(GADM_FILE).exists():
        print(f"[gadm] {GADM_FILE} not found; county-outline overlays will be skipped.")
        return None
    g = gpd.read_file(GADM_FILE)
    sub = g[(g["NAME_1"] == name_1) & (g["NAME_2"] == name_2)].copy()
    if not len(sub):
        print("[gadm] county not found in GADM file; skipping.")
        return None
    sub = to_target(sub)
    print(f"[gadm] {name_2}, {name_1} boundary loaded ({len(sub)} feature).")
    return sub[["GID_2", "NAME_2", "geometry"]]

def add_scalebar(ax, length_mi=5, location=(0.06, 0.04)):
    # Simple scale bar in miles (data units are feet under EPSG:2230).
    x0, x1 = ax.get_xlim(); y0, y1 = ax.get_ylim()
    bar_ft = length_mi * FT_PER_MILE
    bx = x0 + (x1 - x0) * location[0]
    by = y0 + (y1 - y0) * location[1]
    ax.plot([bx, bx + bar_ft], [by, by], color="black", linewidth=3, solid_capstyle="butt", zorder=10)
    ax.text(bx + bar_ft / 2, by + (y1 - y0) * 0.012, f"{length_mi} mi",
            ha="center", va="bottom", fontsize=8, zorder=10)

def add_north_arrow(ax, location=(0.95, 0.92)):
    x0, x1 = ax.get_xlim(); y0, y1 = ax.get_ylim()
    ax.annotate("N", xy=(location[0], location[1]), xytext=(location[0], location[1] - 0.06),
                xycoords="axes fraction", ha="center", va="center", fontsize=11, fontweight="bold",
                arrowprops=dict(arrowstyle="-|>", color="black", lw=1.5), zorder=10)

def style_map_axes(ax, title, source="Sources: US Census tracts/ACS; SANDAG FreightViewer; GADM",
                   scalebar_mi=5, north=True):
    ax.set_title(title, fontweight="bold", fontsize=12, pad=10)
    ax.axis("off")
    if scalebar_mi:
        add_scalebar(ax, length_mi=scalebar_mi)
    if north:
        add_north_arrow(ax)
    ax.text(0.5, -0.02, source, transform=ax.transAxes, ha="center", va="top",
            fontsize=7, color="#555555")

def linear_fit(x, y):
    # Closed-form OLS y = a + b*x with Pearson r and R^2 (numpy only, no extra deps).
    # NaN/inf-safe: non-finite pairs are dropped; degenerate (constant-x or <3 points) returns NaNs.
    x = np.asarray(x, float); y = np.asarray(y, float)
    m = np.isfinite(x) & np.isfinite(y)
    x, y = x[m], y[m]
    if len(x) < 3 or np.ptp(x) == 0 or np.ptp(y) == 0:
        return {"n": int(len(x)), "slope": np.nan, "intercept": np.nan, "r": np.nan, "r2": np.nan}
    b, a = np.polyfit(x, y, 1)
    r = float(np.corrcoef(x, y)[0, 1])
    return {"n": int(len(x)), "slope": float(b), "intercept": float(a), "r": r, "r2": r**2}
```

## 4. ACS industry-of-employment, by tract (resident-based, context)

The 2019-2023 ACS Data Profile (DP03) reports the industry of *employed residents* - that is, where
industrial workers **live**, classified by their industry. This is not the same as the count of
industrial **jobs located** in a tract; for the location question that drives this study, the
workplace-jobs measure in the next subsection (LODES) is primary, and ACS is kept as residential
context. The proxy here is manufacturing + wholesale + transportation/warehousing (NAICS 31-33 /
42 / 48-49).

The employment universe (denominator) is `DP03_0032E`, *civilian employed population 16+*. The
key was already collected in Section 1, so this cell reads it from the environment and does not
prompt again.

```python
# ACS industry-of-employment by tract.
# DP03_0032E civilian employed 16+ (UNIVERSE / denominator)
# DP03_0034E construction (NAICS 23; reported, not in the core proxy)
# DP03_0035E manufacturing (NAICS 31-33)
# DP03_0036E wholesale trade (NAICS 42)
# DP03_0038E transportation & warehousing, and utilities (NAICS 48-49 + 22)

CENSUS_API_KEY = os.environ.get("CENSUS_API_KEY", "").strip()
if not CENSUS_API_KEY:
    raise RuntimeError("CENSUS_API_KEY missing. Re-run the Section 1 setup cell to enter it.")

acs_vars = ["NAME", "DP03_0032E", "DP03_0034E", "DP03_0035E", "DP03_0036E", "DP03_0038E"]
base = f"https://api.census.gov/data/{ACS_YEAR}/acs/acs5/profile"
params = {"get": ",".join(acs_vars), "for": "tract:*",
          "in": "state:06 county:073", "key": CENSUS_API_KEY}

resp = requests.get(base, params=params, timeout=60)
print("Status code:", resp.status_code, "| Content-Type:", resp.headers.get("Content-Type"))

if resp.status_code != 200 or not resp.text.strip().startswith("["):
    raise RuntimeError("Census API request failed.\n\nResponse preview:\n" + resp.text[:800])

rows = resp.json()
if len(rows) <= 1:
    raise RuntimeError("Census API returned no tract-level data.")

acs = pd.DataFrame(rows[1:], columns=rows[0])
acs["GEOID"] = acs["state"] + acs["county"] + acs["tract"]

rename = {
    "DP03_0032E": "emp_total",
    "DP03_0034E": "emp_construction",
    "DP03_0035E": "emp_manufacturing",
    "DP03_0036E": "emp_wholesale",
    "DP03_0038E": "emp_transport_wh_util",
}
acs = acs.rename(columns=rename)
for c in rename.values():
    acs[c] = pd.to_numeric(acs[c], errors="coerce")

# Shared industrial-logistics definition: manufacturing + wholesale + transport/warehousing/utilities
acs["industrial_emp"] = (acs["emp_manufacturing"].fillna(0)
                         + acs["emp_wholesale"].fillna(0)
                         + acs["emp_transport_wh_util"].fillna(0))
acs["industrial_share"] = acs["industrial_emp"] / acs["emp_total"].replace(0, np.nan)

keep = ["GEOID", "NAME", "emp_total", "emp_construction", "emp_manufacturing",
        "emp_wholesale", "emp_transport_wh_util", "industrial_emp", "industrial_share"]
acs = acs[keep].copy()
print("ACS tracts:", acs.shape)
display(acs.head())
```

## 5. Join ACS to the tract geometries

```python
keep = ["GEOID","NAME","emp_total","emp_construction","emp_manufacturing",
        "emp_wholesale","emp_transport_wh_util","industrial_emp","industrial_share"]
tracts = tracts_sd.merge(acs[keep], on="GEOID", how="left")
tracts["industrial_emp_density"] = tracts["industrial_emp"] / tracts["area_sqmi"]
print("Tracts with ACS:", tracts.shape,
      "| missing industrial_emp:", int(tracts["industrial_emp"].isna().sum()))
display(tracts.drop(columns="geometry").head())
```

## 6. Freight infrastructure - ports of entry and major corridors

Ports of entry and major freight roads load from SANDAG FreightViewer, with fallbacks that keep
the analysis runnable if SANDAG is unreachable: POEs fall back to the two San Diego POEs
(San Ysidro, Otay Mesa); major roads fall back to Census TIGER primary/secondary roads.

```python
# ---- Ports of entry (with built-in fallback) -------------------------------
POE_FALLBACK = gpd.GeoDataFrame(
    {"name": ["San Ysidro POE", "Otay Mesa POE"]},
    geometry=[Point(-117.0299, 32.5421), Point(-116.9385, 32.5527)],
    crs="EPSG:4326",
).to_crs(TARGET_CRS)   # approximate coordinates; used only if SANDAG layer fails

try:
    poe = read_esri_or_geojson_url(
        "https://gis.sandag.org/FreightViewer/data/poe_points_update.json",
        "SANDAG POE points")
    if poe is None or len(poe) == 0:
        raise ValueError("empty POE layer")
    poe_source = "SANDAG FreightViewer"
except Exception as e:
    print("[fallback] SANDAG POE unavailable -> built-in POE coordinates:", e)
    poe = POE_FALLBACK.copy()
    poe_source = "built-in fallback (approximate)"
print("POE source:", poe_source, "| n =", len(poe))

# ---- Major roads / freight corridors (with Census fallback) ----------------
roads, roads_source = None, None
try:
    roads = read_esri_or_geojson_url(
        "https://gis.sandag.org/FreightViewer/data/MajorRoads_20171002.json",
        "SANDAG major roads")
    if roads is not None and len(roads):
        roads_source = "SANDAG FreightViewer major roads"
except Exception as e:
    print("[fallback] SANDAG roads unavailable:", e)

if roads is None or len(roads) == 0:
    try:
        roads = gpd.read_file(
            "https://www2.census.gov/geo/tiger/TIGER2023/PRISECROADS/tl_2023_06073_prisecroads.zip")
        roads = to_target(roads)
        if "MTFCC" in roads.columns:   # S1100 primary, S1200 secondary
            roads = roads[roads["MTFCC"].isin(["S1100", "S1200"])].copy()
        roads_source = "Census TIGER primary/secondary roads (fallback)"
    except Exception as e:
        print("[warn] no road layer; corridor indicators will be skipped:", e)

print("Roads source:", roads_source, "| n =", 0 if roads is None else len(roads))
```

## 7. Study areas - Otay Mesa (border) vs. non-border comparisons

Study areas are defined from **City of San Diego Community Plan districts** through a fixed
`CPNAME -> label` map. A tract joins an area when its representative point falls inside that area's
district polygon(s). Otay Mesa is the border case; the other three are non-border industrial
comparisons. If the Community Plan file is unavailable, the assignment falls back to a coordinate +
3-mile-radius window per area, so the section runs without external files.

**Note on the "Sorrento Valley" label:** The City does not maintain a community-plan district named "Sorrento Valley." The Sorrento Valley / Sorrento Mesa industrial-employment area, which is the district intended for comparison, lies within the larger and mostly residential Mira Mesa community-plan district. The label "Sorrento Valley" names that industrial-employment area: under the planning-district definition it is delineated by the enclosing Mira Mesa district, and under the fallback definition by a window centered on the area. The label is applied consistently in the sections that follow.

```python
# City of San Diego Community Plan district -> study-area label.
STUDY_AREA_MAP = {
    "OTAY MESA":             "Otay Mesa",
    "OTAY MESA-NESTOR":      "Otay Mesa",
    "KEARNY MESA":           "Kearny Mesa",
    "MIRAMAR RANCH NORTH":   "Miramar",
    "SCRIPPS MIRAMAR RANCH": "Miramar",
    "MIRA MESA":             "Sorrento Valley",   # Mira Mesa district contains the Sorrento Valley area
}
STUDY_AREAS  = ["Otay Mesa", "Kearny Mesa", "Miramar", "Sorrento Valley"]
BORDER_AREAS = ["Otay Mesa"]

# Coordinate + radius fallback (used only if the Community Plan file is unavailable), so the
# notebook still runs standalone and reproduces the earlier selection.
AREA_CENTERS = {
    "Otay Mesa":       (-117.0548, 32.5751),
    "Kearny Mesa":     (-117.1397, 32.8264),
    "Miramar":         (-117.1050, 32.8970),
    "Sorrento Valley": (-117.2025, 32.8948),
}
AREA_RADIUS_MI = 3.0
CITY_PLAN_URLS = [
    "https://seshat.datasd.org/sde/community_plan/community_plan_datasd.geojson",
    "https://sdgis-sandag.opendata.arcgis.com/datasets/687b4e8f9cc9466cb81e733831992e50_1.geojson",
]

def _load_community_plan():
    p = find_data_file("Community_Plan_SD.geojson")
    if p.exists():
        try:
            g = gpd.read_file(p); print(f"[areas] community plan districts from {p}")
            return g
        except Exception as e:
            print("[areas] local community plan unreadable:", e)
    for url in CITY_PLAN_URLS:
        try:
            g = read_esri_or_geojson_url(url, "Community Plan districts")
            if g is not None and len(g):
                print(f"[areas] community plan districts from {url}")
                return g
        except Exception:
            pass
    return None

area_windows = {}
cp = _load_community_plan()
name_field = None
if cp is not None and len(cp):
    cp = to_target(cp)
    name_field = next((c for c in ["CPNAME", "cpname", "NAME", "name", "COMMUNITY", "community"]
                       if c in cp.columns), None)
if name_field:
    cp["_label"] = cp[name_field].astype(str).str.upper().str.strip().map(STUDY_AREA_MAP)
    for label in STUDY_AREAS:
        sub = cp[cp["_label"] == label]
        if len(sub):
            area_windows[label] = sub.geometry.unary_union
    method = f"Community Plan districts (field '{name_field}')"
else:
    method = "coordinate + 3-mile radius (Community Plan file not found)"

# Any area without a district polygon (or full fallback) uses the coordinate+radius window.
for name, (lon, lat) in AREA_CENTERS.items():
    if name not in area_windows:
        c = gpd.GeoDataFrame(geometry=[Point(lon, lat)], crs="EPSG:4326").to_crs(TARGET_CRS)
        area_windows[name] = c.geometry.iloc[0].buffer(AREA_RADIUS_MI * FT_PER_MILE)

tracts["area"] = "Other"
rep_pts = tracts.geometry.representative_point()
for name in STUDY_AREAS:
    win = area_windows.get(name)
    if win is not None:
        tracts.loc[rep_pts.within(win), "area"] = name

tracts["border"] = np.where(tracts["area"].isin(BORDER_AREAS), "Border (Otay Mesa)",
                    np.where(tracts["area"] == "Other", "Other", "Non-border comparison"))
print("Study-area assignment:", method)
print(tracts["area"].value_counts())
```

## 8. Proximity indicators for every tract

Three geometry-only indicators:
1. `dist_poe_mi` - straight-line distance from the tract's representative point to the nearest POE.
2. `dist_road_mi` - distance to the nearest major freight road.
3. `share_within_1mi_corridor` - fraction of the tract's area within 1 mile of a major road.

These describe tract-level proximity. Establishment-level distances are produced in the upstream
business-points step (Section 9); the hook in Section 10 uses those points only for per-tract counts
and validation, so the two do not duplicate each other.

```python
rep = tracts.geometry.representative_point()

# Distance to nearest POE
poe_u = poe.union_all()
tracts["dist_poe_mi"] = rep.apply(lambda g: g.distance(poe_u)) / FT_PER_MILE

# Road-based indicators (only if a road layer is available)
if roads is not None and len(roads):
    roads_u = roads.union_all()
    tracts["dist_road_mi"] = rep.apply(lambda g: g.distance(roads_u)) / FT_PER_MILE

    buf1 = roads.buffer(1 * FT_PER_MILE).union_all()
    buf_gdf = gpd.GeoDataFrame({"b": [1]}, geometry=[buf1], crs=TARGET_CRS)
    tracts["tract_area"] = tracts.geometry.area
    ov = gpd.overlay(tracts[["GEOID", "geometry", "tract_area"]], buf_gdf, how="intersection")
    ov["ov_area"] = ov.geometry.area
    s = ov.groupby("GEOID", as_index=False)["ov_area"].sum()
    tracts = tracts.drop(columns=[c for c in ["ov_area"] if c in tracts.columns])
    tracts = tracts.merge(s, on="GEOID", how="left")
    tracts["ov_area"] = tracts["ov_area"].fillna(0)
    tracts["share_within_1mi_corridor"] = tracts["ov_area"] / tracts["tract_area"]
else:
    tracts["dist_road_mi"] = np.nan
    tracts["share_within_1mi_corridor"] = np.nan

prev = ["GEOID","area","industrial_emp","industrial_emp_density",
        "dist_poe_mi","dist_road_mi","share_within_1mi_corridor"]
display(tracts[prev].head(8))
```

### 8.1 Workplace jobs by NAICS sector (LODES) - primary activity measure

ACS (Section 4) counts industrial workers by where they *live*. To measure where industrial activity
is actually *located*, this subsection uses the Census LEHD **LODES** Workplace Area Characteristics
(WAC), which count **jobs at the workplace** by NAICS sector. Industrial/logistics workplace jobs are
`CNS05` (manufacturing, 31-33) + `CNS06` (wholesale, 42) + `CNS08` (transportation/warehousing,
48-49) - the same NAICS scope as the rest of the study, but jobs-by-location rather than
residents-by-industry. This is the appropriate measure for a location question, and the per-area
comparison below uses it as the primary activity indicator. If the LODES host is unreachable, the
cell reports that and the notebook falls back to the ACS context.

```python
# LODES8 WAC: jobs located in each block, by NAICS sector; aggregate blocks -> tract.
LODES_STATE = "ca"
LODES_YEARS = [2022, 2021, 2020]                     # newest first
LODES_SEGMENTS = ["CNS05", "CNS06", "CNS08"]         # mfg(31-33), wholesale(42), transport/wh(48-49)

def load_lodes_workplace_industrial():
    for yr in LODES_YEARS:
        url = (f"https://lehd.ces.census.gov/data/lodes/LODES8/{LODES_STATE}/wac/"
               f"{LODES_STATE}_wac_S000_JT00_{yr}.csv.gz")
        dest = DATA_RAW / f"{LODES_STATE}_wac_S000_JT00_{yr}.csv.gz"
        try:
            if not dest.exists():
                r = requests.get(url, timeout=180)
                r.raise_for_status()
                dest.write_bytes(r.content)
            wac = pd.read_csv(dest, compression="gzip",
                              usecols=["w_geocode"] + LODES_SEGMENTS,
                              dtype={"w_geocode": str})
            wac["GEOID"] = wac["w_geocode"].str.zfill(15).str[:11]
            wac["workplace_ind_jobs"] = wac[LODES_SEGMENTS].sum(axis=1)
            out = wac.groupby("GEOID", as_index=False)["workplace_ind_jobs"].sum()
            print(f"[lodes] WAC {yr}: {len(wac):,} blocks -> {len(out):,} tracts")
            return out
        except Exception as e:
            print(f"[lodes] {yr} unavailable: {type(e).__name__}: {e}")
    return None

_lodes = load_lodes_workplace_industrial()
if _lodes is not None:
    tracts = tracts.drop(columns=[c for c in ["workplace_ind_jobs", "workplace_ind_density"]
                                  if c in tracts.columns])
    tracts = tracts.merge(_lodes, on="GEOID", how="left")
    tracts["workplace_ind_jobs"] = tracts["workplace_ind_jobs"].fillna(0)
    tracts["workplace_ind_density"] = tracts["workplace_ind_jobs"] / tracts["area_sqmi"]
else:
    tracts["workplace_ind_jobs"] = np.nan
    tracts["workplace_ind_density"] = np.nan
    print("[lodes] workplace jobs unavailable -> ACS resident employment is used as context only.")
```

```python
# Workplace-based per-area comparison and county-wide association (primary activity measure).
if tracts["workplace_ind_density"].notna().any():
    _study = ["Otay Mesa", "Kearny Mesa", "Miramar", "Sorrento Valley"]
    _wcols = ["workplace_ind_jobs", "workplace_ind_density", "dist_poe_mi", "dist_road_mi"]
    print("Workplace industrial jobs - mean by study area (primary measure):")
    display(tracts[tracts["area"].isin(_study)]
            .groupby("area")[_wcols].mean().reindex(_study).round(2))
    _sub = tracts[["workplace_ind_density", "dist_poe_mi", "dist_road_mi"]].dropna()
    for _x in ["dist_poe_mi", "dist_road_mi"]:
        if _sub[_x].nunique() > 1:
            _r = np.corrcoef(_sub["workplace_ind_density"], _sub[_x])[0, 1]
            print(f"workplace_ind_density ~ {_x}: r = {_r:.3f}")
else:
    print("Workplace jobs not loaded; ACS resident-based indicators (Sections 11-12) used as context.")
```

## 9. Upstream data products

This section prepares three supporting products. Steps 9a and 9b reuse the census tracts and freight
roads already loaded above, so no layer is read twice; each is controlled by a flag and reports a
notice and continues if its inputs are unavailable. Step 9c produces a verification worksheet from
the geocoded points.

* **9a - Geocoded industrial business points.** Reads the City business-tax CSVs, keeps the
  freight-relevant NAICS core (31-33; wholesale 421-424; transport 481-493; 8113; excluding
  425/485/486/487), geocodes the addresses, keeps only matches above a score threshold, assigns each
  point to its tract, and writes `industrial_businesses_core_geocoded.csv` in WGS84. Hook B in
  Section 10 counts these per tract.

  *Geocoding accuracy is not the same as location validity.* A score of 85-100 only indicates how
  confidently the submitted address was matched; it does not establish that the address is the firm's
  warehouse, plant, or operating site rather than a registration or mailing address. The score
  threshold and the `geocode_score` column are kept for transparency, but tract counts from these
  points should be read as counts of *registered* business addresses.

* **9b - Tract-level enrichment via ArcGIS GeoEnrichment (optional).** Produces tract-level business
  and employee counts through `arcgis.geoenrichment.enrich`, converting the tracts to an ArcGIS
  Spatially Enabled DataFrame (SEDF) first. It requires an ArcGIS GeoEnrichment entitlement and is
  disabled by default; the tract-level workplace-jobs measure used in the analysis is the Census
  LODES product from Section 8.1.

* **9c - Address-validity worksheet.** Draws a random sample of the geocoded points and writes a
  worksheet for manual verification of operating locations (see the subsection below). It needs only
  the geocoded file, not an ArcGIS account.

```python
# ---- 9a: geocoded industrial business points ------------------------------
# Output: data_processed/industrial_businesses_core_geocoded.csv (WGS84, with tract_geoid).
# Reuses tracts_sd (tract assignment) and roads (per-business freight distance); nothing re-loaded.
GEN_BUSINESS_POINTS = True
BUSINESS_TAX_FILES  = ["tr_active1.csv", "tr_active2.csv"]
CORE_KEEP = ("31", "32", "33", "421", "422", "423", "424",
             "481", "483", "484", "488", "491", "492", "493", "8113")
CORE_DROP = ("425", "485", "486", "487")

def _naics_group(code):
    s = str(code).strip()
    if s.startswith(("31", "32", "33")): return "Manufacturing"
    if s.startswith("42"):               return "Wholesale Trade"
    if s.startswith(("48", "49")):       return "Transportation and Warehousing"
    if s.startswith("8113"):             return "Industrial Machinery Repair"
    return "Other"

# One ArcGIS connection for the whole notebook: connect on first use, reuse afterward,
# so authentication happens at most once (the username was entered in Section 1).
_ARCGIS_CACHE = {}
def _get_gis():
    if "gis" in _ARCGIS_CACHE:
        return _ARCGIS_CACHE["gis"]
    if not ARCGIS_USERNAME:
        return None
    import arcgis
    from arcgis.gis import GIS
    gis = GIS(username=ARCGIS_USERNAME)
    arcgis.env.active_gis = gis
    _ARCGIS_CACHE["gis"] = gis
    print("ArcGIS connected as:", ARCGIS_USERNAME)
    return gis

def build_business_points():
    files = [f for f in BUSINESS_TAX_FILES if Path(f).exists()]
    if not files:
        print("[9a skip] business-tax CSVs not found; Hook B will use an existing export if present.")
        return
    if not ARCGIS_USERNAME:
        print("[9a skip] no ArcGIS username entered in Section 1; cannot geocode.")
        return
    try:
        from arcgis.geocoding import batch_geocode
        if _get_gis() is None:
            print("[9a skip] ArcGIS not connected.")
            return
    except Exception as e:
        print("[9a skip] ArcGIS unavailable:", type(e).__name__, e)
        return

    df = pd.concat([pd.read_csv(f, dtype=str, engine="python", on_bad_lines="skip") for f in files],
                   ignore_index=True)
    df.columns = (df.columns.str.strip().str.lower()
                  .str.replace(" ", "_").str.replace("#", "num"))
    df["naics"] = df["naics"].astype(str).str.strip().str.replace(".0", "", regex=False)
    core = df[df["naics"].str.startswith(CORE_KEEP, na=False)
              & ~df["naics"].str.startswith(CORE_DROP, na=False)].copy()
    core["industry_group"] = core["naics"].apply(_naics_group)
    for c in ("address", "city", "state", "zip"):
        if c not in core.columns:
            core[c] = ""
    core["full_address"] = (core["address"].fillna("") + ", " + core["city"].fillna("") + ", "
                            + core["state"].fillna("") + " " + core["zip"].fillna(""))
    core = core[core["full_address"].str.len() > 10].reset_index(drop=True)
    if core.empty:
        print("[9a skip] no core industrial addresses to geocode.")
        return

    geo = batch_geocode(core["full_address"].tolist(), out_sr=4326,
                        as_featureset=True).sdf.reset_index(drop=True)
    if len(geo) != len(core):
        print("[9a skip] geocoder row count mismatch; not writing, to avoid misalignment.")
        return
    core["longitude"]    = pd.to_numeric(geo.get("X"), errors="coerce")
    core["latitude"]     = pd.to_numeric(geo.get("Y"), errors="coerce")
    core["geocode_score"] = pd.to_numeric(geo.get("Score"), errors="coerce")
    # Score is geocoding confidence, not proof the address is the operating site (see Section 9 note).
    MIN_GEOCODE_SCORE = 85
    mask = ((geo.get("Status") == "M").values
            & (core["geocode_score"] >= MIN_GEOCODE_SCORE).values
            & core["longitude"].notna().values
            & core["latitude"].notna().values)
    core = core[mask].copy()
    if core.empty:
        print(f"[9a skip] no addresses matched at score >= {MIN_GEOCODE_SCORE}.")
        return

    pts = gpd.GeoDataFrame(core, geometry=gpd.points_from_xy(core["longitude"], core["latitude"]),
                           crs="EPSG:4326").to_crs(TARGET_CRS)
    pts = (gpd.sjoin(pts, tracts_sd[["GEOID", "geometry"]], how="inner", predicate="within")
           .drop(columns="index_right").rename(columns={"GEOID": "tract_geoid"}))
    if roads is not None and len(roads):
        pts = gpd.sjoin_nearest(pts, roads[["geometry"]], how="left",
                                distance_col="dist_to_freight_road_ft")
        if "index_right" in pts.columns:
            pts = pts.drop(columns="index_right")
        pts["dist_to_freight_road_mi"] = pts["dist_to_freight_road_ft"] / FT_PER_MILE

    out = pts.to_crs("EPSG:4326")
    out["longitude"] = out.geometry.x
    out["latitude"]  = out.geometry.y
    cols = [c for c in ["dba_name", "naics", "industry_group", "tract_geoid",
                        "full_address", "dist_to_freight_road_mi", "geocode_score",
                        "longitude", "latitude"] if c in out.columns]
    out_path = DATA_PROCESSED / "industrial_businesses_core_geocoded.csv"
    pd.DataFrame(out)[cols].to_csv(out_path, index=False)
    print(f"[9a ok] wrote {out_path.name}: {len(out):,} businesses, "
          f"{out['tract_geoid'].nunique()} tracts")

if GEN_BUSINESS_POINTS:
    build_business_points()
```

```python
# ---- 9b: per-tract job counts via ESRI GeoEnrichment (OPTIONAL cross-check) -----
# The tract-level workplace-jobs measure used in the analysis is LODES (Section 8.1); this step is
# OFF by default and runs only to cross-check ESRI's 2-digit NAICS job counts (needs an entitlement).
GEN_ENRICHMENT = False
ENRICH_VARS = ["businesses.N06", "businesses.N07", "businesses.N21",
               "employees.N06", "employees.N07", "employees.N21"]

def build_enrichment():
    if not ARCGIS_USERNAME:
        print("[9b skip] no ArcGIS username entered in Section 1; cannot enrich.")
        return
    try:
        from arcgis.geoenrichment import enrich      # correct module (not arcgis.enrichment)
        from arcgis.features import GeoAccessor       # SEDF conversion
        if _get_gis() is None:
            print("[9b skip] ArcGIS not connected.")
            return
    except Exception as e:
        print("[9b skip] ArcGIS GeoEnrichment unavailable:", type(e).__name__, e)
        return

    # ESRI needs a Spatially Enabled DataFrame (SEDF), not a GeoPandas GeoDataFrame.
    src_gdf = tracts_sd[["GEOID", "geometry"]].to_crs("EPSG:4326").copy()
    try:
        src_sedf = GeoAccessor.from_geodataframe(src_gdf)
    except Exception as e:
        print("[9b skip] could not convert tracts to SEDF:", type(e).__name__, e)
        return
    try:
        enriched = enrich(study_areas=src_sedf, analysis_variables=ENRICH_VARS)
    except Exception as e:
        print("[9b skip] enrich() call failed:", type(e).__name__, e)
        return
    if enriched is None or not len(enriched):
        print("[9b skip] enrichment returned no rows.")
        return

    out = pd.DataFrame(enriched)
    if "GEOID" not in out.columns and len(out) == len(src_gdf):
        out["GEOID"] = src_gdf["GEOID"].values
    keep = ["GEOID"] + [c for c in out.columns
                        if c != "GEOID" and pd.api.types.is_numeric_dtype(out[c])
                        and any(k in c.upper() for k in ("N06", "N07", "N21", "BUS", "EMP"))]
    if len(keep) <= 1:
        keep = ["GEOID"] + [c for c in out.columns
                            if c != "GEOID" and pd.api.types.is_numeric_dtype(out[c])]
    out_path = DATA_PROCESSED / "esri_enrichment_by_tract.csv"
    out[keep].to_csv(out_path, index=False)
    print(f"[9b ok] wrote {out_path.name}: {len(out):,} tracts, {len(keep) - 1} variables")

if GEN_ENRICHMENT:
    build_enrichment()
else:
    print("[9b skip] ESRI GeoEnrichment is disabled. Tract-level workplace jobs are taken from "
          "Census LODES (Section 8.1), the activity measure used throughout; this step remains "
          "available only as an optional cross-check where an ArcGIS GeoEnrichment entitlement exists.")
```

### 9c. Address-validity check (sample worksheet)

A geocoding score of 85+ only means the submitted address was matched with high *confidence*; it
does **not** establish that the listed address is the firm's actual warehouse, plant, or operating
site rather than a mailing address or headquarters. To check this rather than assume it, the cell
below draws a small random sample of geocoded businesses and writes a worksheet
(`tables/address_validity_worksheet.csv`) with the automatable fields pre-filled and three columns
left blank. Complete those columns by hand: look up each firm's real operating location (company
website, maps, directory), record it, and mark whether it matches the tax-listing address. The
result feeds the geocoding caveat in the limitations (Section 16): the point counts are of
*registered addresses*, with whatever share is found to be mismatched.

```python
# Draw a sample to verify by hand. Geocoding score = match confidence, NOT operating-site validity.
VALIDITY_SAMPLE_N = 12
_biz_path = find_data_file("industrial_businesses_core_geocoded.csv")
if _biz_path.exists():
    _vdf = pd.read_csv(_biz_path, dtype=str)
    name_col = next((c for c in ["dba_name","name","business_name","owner_name"] if c in _vdf.columns), None)
    addr_col = next((c for c in ["full_address","address","situs_address"] if c in _vdf.columns), None)
    n = min(VALIDITY_SAMPLE_N, len(_vdf))
    sample = _vdf.sample(n=n, random_state=447).reset_index(drop=True)
    work = pd.DataFrame({
        "business":            sample[name_col] if name_col else "(name not in file)",
        "naics":               sample.get("naics", ""),
        "tax_listing_address": sample[addr_col] if addr_col else "(address not saved in geocoded file)",
        "geocoder_matched_address": sample.get("matched_address", ""),
        "planning_area":       sample.get("planning_area", ""),
        "geocoded_tract":      sample.get("tract_geoid", ""),
        "geocode_score":       sample.get("geocode_score", ""),
        "geocoded_lon":        sample.get("longitude", ""),
        "geocoded_lat":        sample.get("latitude", ""),
        # ---- fill these THREE columns MANUALLY after checking each firm's website / maps ----
        "operating_site_found": "",   # the actual plant/warehouse address you find
        "address_matches":      "",   # yes / no / partial
        "notes":                "",   # e.g. "listing is HQ; plant is elsewhere"
    })
    vout = TABLES / "address_validity_worksheet.csv"
    work.to_csv(vout, index=False)
    print(f"[9c] wrote {vout.name}: {len(work)} businesses to verify by hand.")
    print("     Score is match CONFIDENCE, not proof the address is the operating site.")
    display(work)
else:
    print("[9c skip] no geocoded business file yet; run Section 9a or provide the geocoded CSV.")
```

## 10. Integration hooks

Each hook looks for an optional input file in `data_processed/` or the working directory, and
continues with a printed notice if the file is absent, so the section completes on the public data
already loaded.

* **Hook A - external tract-level job counts (optional cross-check).** If a tract-level LODES jobs
  file (`lodes_enrichment_by_tract.csv`) is present, its `lodes_industrial_jobs` column is correlated
  with this notebook's `workplace_ind_jobs` (Section 8.1) as a consistency check; Section 8.1 remains
  the source used in the analysis. An Esri GeoEnrichment file (`esri_enrichment_by_tract.csv`) is
  merged only if `USE_ESRI_ENRICHMENT` is on and the file exists.
* **Hook B - geocoded business points.** Counts establishments per tract from
  `industrial_businesses_core_geocoded.csv` (WGS84 `longitude`/`latitude`) and reports their
  correlation with ACS industrial employment. If the file carries a `tract_geoid` column the counts
  use it directly; otherwise the points are spatially joined here. The hook only counts and
  validates - it does not recompute establishment-level distances.

```python
# Hook A - external tract-level job counts cross-check (optional); optional Esri merge.
# Files may live in data_processed/ or the working directory; find_data_file checks both.
lodes_x_path = find_data_file("lodes_enrichment_by_tract.csv")
if lodes_x_path.exists():
    jx = pd.read_csv(lodes_x_path, dtype={"GEOID": str})
    jx.columns = [c.strip() for c in jx.columns]
    jcol = next((c for c in ["lodes_industrial_jobs", "lodes_ind_jobs"] if c in jx.columns), None)
    if jcol and "workplace_ind_jobs" in tracts.columns:
        chk = (tracts[["GEOID", "workplace_ind_jobs"]]
               .merge(jx[["GEOID", jcol]], on="GEOID", how="inner"))
        fit = linear_fit(chk["workplace_ind_jobs"], chk[jcol])
        print(f"[hook A] external LODES cross-check ({jcol}): r = {fit['r']:.3f} (n={fit['n']}); "
              f"Section 8.1 is the source used in the analysis.")
    else:
        print("[hook A] external LODES file found but expected columns missing; cross-check skipped.")
else:
    print("[hook A] no external LODES file (lodes_enrichment_by_tract.csv) -> using Section 8.1 only.")

# Optional Esri GeoEnrichment merge (normally absent; LODES is the measure used).
esri_path = find_data_file("esri_enrichment_by_tract.csv")
if USE_ESRI_ENRICHMENT and esri_path.exists():
    enr = pd.read_csv(esri_path, dtype={"GEOID": str})
    enr.columns = [c.strip() for c in enr.columns]
    new_cols = [c for c in enr.columns if c != "GEOID"]
    tracts = tracts.drop(columns=[c for c in new_cols if c in tracts.columns])
    tracts = tracts.merge(enr, on="GEOID", how="left")
    print("[hook A] merged optional Esri enrichment columns:", new_cols)
elif USE_ESRI_ENRICHMENT:
    print(f"[hook A] Esri enabled, but {esri_path.name} not found -> continuing without it.")
```

```python
# Hook B - geocoded business points, counted per tract
biz_path = find_data_file("industrial_businesses_core_geocoded.csv")
if USE_BUSINESS_POINTS and biz_path.exists():
    bdf = pd.read_csv(biz_path)
    if "tract_geoid" in bdf.columns:
        # The geocoded output already assigned each point to a tract.
        bdf["tract_geoid"] = bdf["tract_geoid"].astype(str)
        cnt = (bdf.groupby("tract_geoid").size()
                  .rename("business_points").reset_index()
                  .rename(columns={"tract_geoid": "GEOID"}))
        method = "tract_geoid column"
    else:
        bpts = gpd.GeoDataFrame(
            bdf, geometry=gpd.points_from_xy(bdf["longitude"], bdf["latitude"]),
            crs="EPSG:4326").to_crs(TARGET_CRS)
        joined = gpd.sjoin(bpts, tracts[["GEOID", "geometry"]], how="inner", predicate="within")
        cnt = joined.groupby("GEOID").size().rename("business_points").reset_index()
        method = "spatial join"
    cnt["GEOID"] = cnt["GEOID"].astype(str)
    tracts = tracts.drop(columns=[c for c in ["business_points"] if c in tracts.columns])
    tracts = tracts.merge(cnt, on="GEOID", how="left")
    tracts["business_points"] = tracts["business_points"].fillna(0)
    fit = linear_fit(tracts["business_points"], tracts["industrial_emp"])
    print(f"[hook B] counted via {method}; points vs ACS industrial employment: "
          f"r = {fit['r']:.2f} (n={fit['n']})")
elif USE_BUSINESS_POINTS:
    print(f"[hook B] enabled, but {biz_path} not found -> continuing without point counts.")
else:
    print("[hook B] disabled.")
```

### 13.1 Border-logistics indicator businesses (composition and index)

Two-digit NAICS sectors may be too coarse to separate a border-logistics economy from general
industrial activity, so this subsection uses finer codes from the geocoded business points
(Section 9a / Hook B) to identify *indicator* businesses for the near-border economy: freight
forwarding and customs brokerage (NAICS 4885 / 488), warehousing and storage (493), truck
transportation (484), and couriers (492). For each study area it reports the share of industrial
businesses that are indicator businesses (composition) and the share of all such indicator businesses
located in that area (concentration), forming a simple **border-logistics indicator index**. This
addresses whether Otay Mesa differs from Kearny Mesa in *composition* rather than only in proximity.

These code groups are chosen because their business model depends on moving goods across the border
or staging them near it: customs brokers and freight forwarders clear cross-border shipments,
warehouses stage them, and truckers and couriers carry them, so a high indicator share signals a
border-logistics orientation rather than generic manufacturing. *Granularity caveat:* the tax-listing
NAICS are matched by prefix at whatever depth the source provides, and 488 is a broad parent that also
covers non-forwarding support activities, so the index is indicative rather than exact.
It runs only if the business-points file exists; not every freight-related measure is expected to
single out Otay Mesa, and that is itself a result worth reporting.

```python
biz_path = find_data_file("industrial_businesses_core_geocoded.csv")
# Documented border-logistics indicator NAICS (prefix -> what it captures, with 6-digit subtypes).
# Matching is by PREFIX, which already subsumes the finer codes (e.g. "493110".startswith("493")).
# "488" is broad (support activities for transportation) and is retained for continuity; its
# border-specific subtype is 4885. If the source carried only 6-digit codes the same matches hold.
BORDER_LOGISTICS_NAICS = {
    "488":  "Support activities for transportation (broad; includes 4885 below)",
    "4885": "Freight transportation arrangement / forwarding / customs brokerage (e.g. 488510)",
    "493":  "Warehousing and storage (e.g. 493110 general warehousing)",
    "484":  "Truck transportation (e.g. 484121/484122 long-distance general freight)",
    "492":  "Couriers and messengers / express delivery (e.g. 492110)",
}
INDICATOR_PREFIXES = tuple(BORDER_LOGISTICS_NAICS)   # forwarding/customs, warehousing, trucking, couriers
print("Border-logistics indicator NAICS (prefix -> meaning):")
for _k, _v in BORDER_LOGISTICS_NAICS.items():
    print(f"  {_k:5s} {_v}")

if biz_path.exists():
    bdf = pd.read_csv(biz_path, dtype={"naics": str, "tract_geoid": str})
    bdf["naics"] = bdf["naics"].fillna("").str.strip()
    # Granularity diagnostic: shows the depth of NAICS actually present in the source, so the
    # limitation statement ("source granularity limits finer classification") is empirical.
    _len_dist = bdf.loc[bdf["naics"] != "", "naics"].str.len().value_counts().sort_index()
    print("NAICS code-length distribution (chars -> count):", dict(_len_dist))
    bdf["is_indicator"] = bdf["naics"].str.startswith(INDICATOR_PREFIXES)
    # map each business to a study area via its tract
    area_lookup = tracts[["GEOID", "area"]].drop_duplicates()
    if "tract_geoid" in bdf.columns:
        bdf = bdf.merge(area_lookup, left_on="tract_geoid", right_on="GEOID", how="left")
    else:
        pts = gpd.GeoDataFrame(bdf, geometry=gpd.points_from_xy(bdf["longitude"], bdf["latitude"]),
                               crs="EPSG:4326").to_crs(TARGET_CRS)
        pts = gpd.sjoin(pts, tracts[["GEOID", "area", "geometry"]], how="left", predicate="within")
        bdf = pd.DataFrame(pts.drop(columns="geometry"))
    bdf["area"] = bdf["area"].fillna("Other")

    total_ind = int(bdf["is_indicator"].sum())
    rows = []
    for a in ["Otay Mesa", "Kearny Mesa", "Miramar", "Sorrento Valley"]:
        sa = bdf[bdf["area"] == a]
        n_all = len(sa); n_ind = int(sa["is_indicator"].sum())
        rows.append({"area": a, "n_industrial": n_all, "n_indicator": n_ind,
                     "indicator_share": (n_ind / n_all if n_all else np.nan),
                     "pct_of_all_indicators": (n_ind / total_ind if total_ind else np.nan)})
    idx_tbl = pd.DataFrame(rows).set_index("area").round(3)
    print("Border-logistics indicator businesses by area:")
    display(idx_tbl)
    idx_tbl.to_csv(TABLES / "border_logistics_indicator_index.csv")
    if idx_tbl.loc["Otay Mesa", "n_industrial"] and idx_tbl.loc["Kearny Mesa", "n_industrial"]:
        print(f"Indicator share - Otay Mesa {idx_tbl.loc['Otay Mesa','indicator_share']:.3f} "
              f"vs Kearny Mesa {idx_tbl.loc['Kearny Mesa','indicator_share']:.3f}")
else:
    print(f"[13.1 skip] {biz_path.name} not found; run Section 9a (business points) to enable this.")
```

## 15. Overlap measures - industrial zoning, general-plan land use, employment centers, and freight infrastructure

This section measures how four planning and infrastructure layers coincide in space, region-wide
and inside each study area.

**Layers.**
- *Industrial zoning* - City of San Diego zoning (codes IP / IL / IH / IS / IBT). Source:
  https://data.sandiego.gov/datasets/zoning/
- *Industrial general-plan land use* - City of San Diego General Plan, industrial / employment
  categories. Source: https://data.sandiego.gov/datasets/general-plan-land-use/
- *Employment centers* - SANDAG Employment Centers. Source:
  https://opendata.sandag.org/stories/s/Employment-Centers-V2-Landing-Page/grty-wn99/
- *Freight infrastructure* - ports of entry and major roads loaded above, optionally extended with
  additional FreightViewer layers, expressed as a 1-mile freight-corridor buffer for areal overlap.

**Definition.** "Industrially relevant land" is industrial zoning combined with industrial land
use (union); the zoning vs land-use agreement is reported separately.

**Measures.** For each layer pair: intersection area, the Jaccard index (area-intersection /
area-union) and containment (area-intersection / area-A). A 4-way coupling core
(relevant land intersect employment centers intersect freight corridor) is computed, and all
measures are aggregated to the census tract so they join the per-tract table.

Each loader has a fallback; a portal outage reduces the section gracefully rather than stopping
the notebook. High overlap indicates spatial coincidence, not a causal relationship.

```python
# ---- Section 15: load the four layer families ------------------------------
CORRIDOR_BUFFER_MI = 1.0
IND_ZONE_PREFIXES = ("IP", "IL", "IH", "IS", "IBT")
IND_LU_KEYWORDS = ("industrial", "manufactur", "warehouse", "business park")

ZONING_URLS  = ["https://geo.sandag.org/server/rest/directories/downloads/Zoning_Base_SD.geojson"]
LANDUSE_URLS = [
    # SANDAG downloads directory - filename matches their portal item "General_Plan_Land_Use_SD"
    "https://geo.sandag.org/server/rest/directories/downloads/General_Plan_Land_Use_SD.geojson",
    "https://geo.sandag.org/server/rest/directories/downloads/GeneralPlan_LandUse_SD.geojson",
]
EMP_URLS = [
    # SANDAG ArcGIS REST - layer 0 is "Employment Center Boundaries" (polygons); Esri JSON -> WGS84
    "https://gis.sandag.org/sdgis/rest/services/RegionalVision/5BM_Employment_CombinedECs_HNTB/MapServer/0/query?where=1%3D1&outFields=*&returnGeometry=true&outSR=4326&f=json",
    "https://gis.sandag.org/sdgis/rest/services/RegionalVision/5BM_Employment/MapServer/0/query?where=1%3D1&outFields=*&returnGeometry=true&outSR=4326&f=json",
    "https://opendata.sandag.org/resource/grty-wn99.geojson?$limit=5000",
]
EXTRA_FREIGHT_URLS = {
    "rail_yards": "https://gis.sandag.org/FreightViewer/data/rail_yards_poly_dom.json",
    "marine":     "https://gis.sandag.org/FreightViewer/data/MarineTerminals.json",
}

def _load_first(urls, name):
    for u in urls:
        try:
            g = read_esri_or_geojson_url(u, name)
            if g is not None and len(g):
                return g
        except Exception as e:
            print(f"[..] {name}: {type(e).__name__} <- {u}")
    print(f"[warn] {name}: unavailable -> skipped")
    return None

def _pick_col(g, keys):
    return next((c for c in g.columns if any(k in c.lower() for k in keys)), None)

# industrial zoning
ind_zone = None
_z = _load_first(ZONING_URLS, "City zoning")
if _z is not None:
    _zc = _pick_col(_z, ["zone"])
    if _zc:
        ind_zone = _z[_z[_zc].astype(str).str.upper().str.strip()
                      .str.startswith(IND_ZONE_PREFIXES)][["geometry"]].copy()
        print("industrial zoning polygons:", len(ind_zone), f"(col '{_zc}')")

# industrial / employment general-plan land use
ind_lu = None
_lu = _load_first(LANDUSE_URLS, "General Plan land use")
if _lu is not None:
    _lc = _pick_col(_lu, ["landuse", "land_use", "lu_", "desc", "category", "type", "gplu"])
    if _lc:
        _m = _lu[_lc].astype(str).str.lower().str.contains("|".join(IND_LU_KEYWORDS), na=False)
        ind_lu = _lu[_m][["geometry"]].copy()
        print("industrial land-use polygons:", len(ind_lu),
              "| categories:", sorted(_lu.loc[_m, _lc].astype(str).unique())[:10])

# employment centers
emp_centers = None
_ec = _load_first(EMP_URLS, "SANDAG employment centers")
if _ec is not None:
    emp_centers = _ec[["geometry"]].copy()

# freight infrastructure -> 1-mile corridor buffer (reuse POE + roads, add extras if available)
from shapely.ops import unary_union
freight_src_layers = []                      # raw layers, kept for the distance-sensitivity check
freight_geoms = []
for _g in (poe, roads):
    if _g is not None and len(_g):
        freight_src_layers.append(_g[["geometry"]].copy())
        freight_geoms.append(_g.buffer(CORRIDOR_BUFFER_MI * FT_PER_MILE).union_all())
for _k, _u in EXTRA_FREIGHT_URLS.items():
    try:
        _g = read_esri_or_geojson_url(_u, f"freight/{_k}")
        if _g is not None and len(_g):
            freight_src_layers.append(_g[["geometry"]].copy())
            freight_geoms.append(_g.buffer(CORRIDOR_BUFFER_MI * FT_PER_MILE).union_all())
    except Exception:
        pass
freight_buffer = (gpd.GeoDataFrame({"name": ["freight_corridor"]},
                                   geometry=[unary_union(freight_geoms)], crs=TARGET_CRS)
                  if freight_geoms else None)

print("layers ready:",
      {"zoning": ind_zone is not None, "landuse": ind_lu is not None,
       "emp_centers": emp_centers is not None, "freight_buffer": freight_buffer is not None})
```

```python
# ---- Section 15: relevant land + pairwise overlap measures -----------------
def _ua(g):
    return g.union_all()

def industrially_relevant_land(zoning, landuse):
    parts = [g for g in (zoning, landuse) if g is not None and len(g)]
    if not parts:
        return None
    geom = gpd.GeoDataFrame(pd.concat(parts, ignore_index=True), crs=TARGET_CRS).union_all()
    return gpd.GeoDataFrame({"name": ["relevant_land"]}, geometry=[geom], crs=TARGET_CRS)

def pairwise_overlap(a, b, na, nb, clip_to=None):
    if a is None or b is None or not len(a) or not len(b):
        return None
    ag, bg = _ua(a), _ua(b)
    if clip_to is not None:
        ag, bg = ag.intersection(clip_to), bg.intersection(clip_to)
    inter, union, aa, ab = ag.intersection(bg).area, ag.union(bg).area, ag.area, bg.area
    return {"layer_A": na, "layer_B": nb,
            "area_A_sqmi": aa / FT_PER_MILE**2, "area_B_sqmi": ab / FT_PER_MILE**2,
            "intersection_sqmi": inter / FT_PER_MILE**2,
            "jaccard": (inter / union) if union else np.nan,
            "containment_A_in_B": (inter / aa) if aa else np.nan,
            "containment_B_in_A": (inter / ab) if ab else np.nan}

relevant_land = industrially_relevant_land(ind_zone, ind_lu)

zlu = pairwise_overlap(ind_zone, ind_lu, "industrial_zoning", "industrial_landuse")
if zlu:
    print(f"Zoning vs land-use agreement: Jaccard={zlu['jaccard']:.3f} | "
          f"zoning inside land use={zlu['containment_A_in_B']:.3f}")
else:
    print("Zoning/land-use agreement not computable (a layer is missing).")

polys = {}
if relevant_land  is not None: polys["relevant_land"] = relevant_land
if emp_centers    is not None: polys["employment_centers"] = emp_centers
if freight_buffer is not None: polys["freight_corridor"] = freight_buffer
if ind_zone       is not None: polys["industrial_zoning"] = ind_zone
if ind_lu         is not None: polys["industrial_landuse"] = ind_lu

_rows = []
_names = list(polys)
_scopes = {"_REGION_": None}
_scopes.update(area_windows)
for _area, _win in _scopes.items():
    for _i in range(len(_names)):
        for _j in range(_i + 1, len(_names)):
            _m = pairwise_overlap(polys[_names[_i]], polys[_names[_j]],
                                  _names[_i], _names[_j], clip_to=_win)
            if _m:
                _rows.append({"area": _area, **_m})
overlap_tbl = pd.DataFrame(_rows)
if len(overlap_tbl):
    overlap_tbl.to_csv(TABLES / "overlap_pairwise_by_area.csv", index=False)
    _show = overlap_tbl[overlap_tbl["area"].isin(["_REGION_"] + BORDER_AREAS)]
    print("Pairwise overlap (region + Otay Mesa):")
    display(_show[["area", "layer_A", "layer_B", "jaccard", "containment_A_in_B"]].round(3))
else:
    print("No polygon layers available -> overlap table empty (tract analysis is unaffected).")
```

```python
# ---- Section 15: 4-way coupling core + aggregate overlaps to tracts --------
def coupling_core(irl, emp, fbuf):
    parts = [g for g in (irl, emp, fbuf) if g is not None and len(g)]
    if len(parts) < 2:
        return None
    geom = _ua(parts[0])
    for g in parts[1:]:
        geom = geom.intersection(_ua(g))
    return None if geom.is_empty else gpd.GeoDataFrame(
        {"name": ["coupling_core"]}, geometry=[geom], crs=TARGET_CRS)

def share_per_tract(tr, layer):
    if layer is None or not len(layer):
        return pd.Series(np.nan, index=tr.index)
    lyr = gpd.GeoDataFrame({"_k": [1]}, geometry=[_ua(layer)], crs=TARGET_CRS)
    base = tr[["GEOID", "geometry"]].copy()
    base["ta"] = base.geometry.area
    ov = gpd.overlay(base, lyr, how="intersection")
    if not len(ov):
        return pd.Series(0.0, index=tr.index)
    ov["oa"] = ov.geometry.area
    s = ov.groupby("GEOID")["oa"].sum()
    mg = base.merge(s.rename("a"), on="GEOID", how="left")
    mg["a"] = mg["a"].fillna(0.0)
    return pd.Series((mg["a"] / mg["ta"]).values, index=tr.index)

core = coupling_core(relevant_land, emp_centers, freight_buffer)

tracts["share_ind_zoning"]       = share_per_tract(tracts, ind_zone).values
tracts["share_ind_landuse"]      = share_per_tract(tracts, ind_lu).values
tracts["share_relevant_land"]    = share_per_tract(tracts, relevant_land).values
tracts["share_emp_center"]       = share_per_tract(tracts, emp_centers).values
tracts["share_freight_corridor"] = share_per_tract(tracts, freight_buffer).values
tracts["share_coupling_core"]    = share_per_tract(tracts, core).values

ov_cols = ["share_ind_zoning", "share_ind_landuse", "share_relevant_land",
           "share_emp_center", "share_freight_corridor", "share_coupling_core"]
tracts[["GEOID", "area"] + ov_cols].to_csv(TABLES / "overlap_shares_by_tract.csv", index=False)

print("Per-tract overlap shares added. Mean share by study area:")
display(tracts[tracts["area"] != "Other"].groupby("area")[ov_cols].mean().round(3))
```

### 15.1 Sensitivity to the corridor distance

The 1-mile corridor is one arbitrary choice. This subsection recomputes the share of each study
area's industrially relevant land that falls inside the freight corridor at several buffer
distances (0.5, 1, and 2 miles), so the comparison between areas does not hinge on a single radius.
A stable ranking across distances indicates the finding is not an artifact of the chosen buffer.

```python
# Rebuild the freight corridor at several distances and re-measure containment of relevant land.
if relevant_land is not None and freight_src_layers and area_windows:
    _rl_geom = relevant_land.geometry.iloc[0]
    _dists = [0.5, 1.0, 2.0]
    _areas = ["Otay Mesa", "Kearny Mesa", "Miramar", "Sorrento Valley"]
    _rows = []
    for d in _dists:
        bufs = [lyr.buffer(d * FT_PER_MILE).union_all() for lyr in freight_src_layers]
        corridor = unary_union(bufs)
        row = {"corridor_mi": d}
        for a in _areas:
            win = area_windows.get(a)
            if win is None:
                row[a] = np.nan; continue
            rl_a = _rl_geom.intersection(win)
            row[a] = (rl_a.intersection(corridor).area / rl_a.area) if rl_a.area else np.nan
        _rows.append(row)
    sens = pd.DataFrame(_rows).set_index("corridor_mi").round(3)
    print("Share of industrially relevant land inside the freight corridor, by distance:")
    display(sens)
    sens.to_csv(TABLES / "sensitivity_corridor_distance.csv")
else:
    print("[15.1 skip] need relevant_land + freight sources + study windows for the sensitivity check.")
```

## 16. Interpretation and limitations

**Reading the results.** The evidence gives a qualified, mostly negative answer to the question of
whether border-zone industrial activity is *distinctively* more freight-coupled than comparable
inland industrial areas. Otay Mesa stands out on only two measures - its proximity to the ports of
entry (close to definitional, since the district sits on the border) and the *composition* of its
industrial base - while on the scale, density, and spatial-coverage measures it is comparable to, or
below, the inland comparison areas.

1. *Activity and proximity (Sections 8.1, 11-13).* Workplace industrial jobs (LODES WAC) are the
   primary activity measure; ACS industry-of-employment is reported only as residential context. By
   workplace jobs, Otay Mesa is the *smallest* of the four areas (about 131 industrial jobs per tract,
   vs. 811 in Kearny Mesa and 1,595 in Sorrento Valley) and only middling by density (about 203/sq mi -
   above Kearny and Miramar, well below Sorrento Valley's 508). The county-wide association between
   workplace industrial density and freight distance is essentially zero (r = -0.024 to ports of entry,
   r = -0.068 to freight roads). The ACS resident-based density shows the expected negative sign but is
   weak (r = -0.18 to POE, -0.22 to roads; R-squared about 0.03-0.05), and Otay Mesa's resident-based
   density (about 835/sq mi) is indistinguishable from the non-border comparison mean (about 851).
   Otay's one clear proximity advantage - about 3.4 mi to the nearest POE vs. about 24 mi for the
   comparison areas - is near-definitional.
2. *Spatial overlap (Section 15).* Across the county, about 82% of all industrially relevant land lies
   within one mile of freight infrastructure, so corridor adjacency is a *general* feature of San Diego
   industrial land rather than an Otay-specific one. Otay Mesa's relevant land is almost entirely inside
   the corridor (about 99.9%), but Kearny Mesa is effectively tied (about 99.6%), so containment does
   not distinguish the border. Per tract, Otay actually has the *lowest* industrially relevant land
   share (about 0.020) and the *lowest* coupling-core share (about 0.020) of the four areas: its high
   containment reflects that the little relevant land it has happens to sit beside border freight
   infrastructure, not that it has more. The two definitions of industrial land (zoning vs. general-plan
   land use) agree only modestly (region Jaccard about 0.30; Otay about 0.26), so "relevant land" is
   sensitive to definition.
3. *Composition (Section 13.1).* This is the one measure on which Otay Mesa is genuinely distinctive:
   about 27.8% of its industrial businesses fall in the border-logistics indicator NAICS (freight
   forwarding / customs brokerage, warehousing, trucking, couriers) - the highest share of the four
   areas (Kearny about 17.9%, Miramar about 24.2%, Sorrento about 20.3%). Its industrial base is
   therefore the most logistics-*specialized* in composition. However, the absolute count is small
   (about 59 indicator businesses) and Otay holds only about 7.3% of the region's indicator businesses
   (Sorrento holds about 11.9%), so specialization does not translate into scale.
4. *Robustness (Section 15.1).* The corridor-containment ranking is stable across 0.5-, 1-, and 2-mile
   buffers (Otay about Kearny at the top, then Sorrento Valley, then Miramar), so the comparison does not
   hinge on the arbitrary 1-mile radius. At every distance Otay Mesa is at most marginally above Kearny
   Mesa, reinforcing that the border area is not distinctively more corridor-coupled.

**Interpretation.** Taken together, the indicators describe Otay Mesa as broadly similar to the
established inland industrial districts in industrial scale, employment density, and freight-corridor
coverage, and as distinct mainly in two respects: its proximity to the ports of entry, which largely
follows from its location on the international boundary, and the composition of its industrial base,
which is more concentrated in border-logistics activities. Corridor proximity and land-use overlap
are characteristic of industrial land throughout the county rather than specific to the border zone.
These relationships are spatial associations measured across census tracts; they describe the
co-location of industrial activity and freight infrastructure but do not establish that the
infrastructure caused industrial development. That several freight-related indicators do not
distinguish the border zone is itself an informative result, suggesting that the industrial geography
of the area is only partly explained by freight access.

**Limitations.**
- *Workplace vs. resident measures.* The conceptually correct workplace measure (LODES) and the
  residential ACS measure tell slightly different stories; the workplace measure - the appropriate one
  for a location question - shows the weakest freight association. If LODES did not load in a given run,
  only the ACS residential context is available and is labeled as such.
- *Missing employment-centers layer.* The SANDAG employment-center boundaries could not be loaded (the
  service requires authentication), so `share_emp_center` is undefined and the coupling core is computed
  as relevant land intersect freight corridor (three layers: zoning, land use, freight) rather than the
  full four-way intersection. This understates one dimension of coupling and is reported as such.
- *Definition sensitivity.* Zoning and land-use definitions of industrial land overlap only modestly
  (Jaccard about 0.3), so the industrially relevant footprint depends on the chosen definition.
- *Geocoding accuracy is not location validity.* Business-point scores reflect match confidence, not
  that an address is an operating site rather than a mailing/registration address. Section 9c samples
  businesses into an `address_validity_worksheet.csv` for manual verification; report the share found to
  be mismatched, and treat the point counts as counts of registered addresses pending that check.
- *Land-allocation confounding.* Otay Mesa's high corridor containment partly reflects that it has
  little industrial land overall, all of it adjacent to border infrastructure, rather than an
  independent pull of industry toward freight.
- *Exploratory, not causal.* Straight-line distance, circular study windows, US-side-only data, and the
  2020-vintage tracts are simplifications; tract counts per area are small and the OLS is exploratory,
  not a causal model.
